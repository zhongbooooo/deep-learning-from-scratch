# Class 2: Neural Network Training from Scratch

This notebook implements a very small neural network step by step to understand:

- Forward pass
- ReLU activation
- Loss function
- Backpropagation
- Gradient descent

```python
x1 = 1
x2 = 2

w1 = 0.5
w2 = -0.2
b = 0.1

z = w1 * x1 + w2 * x2 + b

print("z =", z)
```

```python
def relu(z):
    return max(0, z)

y_pred = relu(z)

print("prediction =", y_pred)
```

```python
y = 1

loss = 0.5 * (y - y_pred) ** 2

print("loss =", loss)
```

```python
# Gradients
dL_dy = y_pred - y

dy_dz = 1 if z > 0 else 0

dz_dw1 = x1
dz_dw2 = x2
dz_db = 1

grad_w1 = dL_dy * dy_dz * dz_dw1
grad_w2 = dL_dy * dy_dz * dz_dw2
grad_b = dL_dy * dy_dz * dz_db

print("grad_w1 =", grad_w1)
print("grad_w2 =", grad_w2)
print("grad_b =", grad_b)
```

```python
learning_rate = 0.1

w1 = w1 - learning_rate * grad_w1
w2 = w2 - learning_rate * grad_w2
b = b - learning_rate * grad_b

print("new w1 =", w1)
print("new w2 =", w2)
print("new b =", b)
```

```python
z = w1 * x1 + w2 * x2 + b
y_pred = relu(z)

loss = 0.5 * (y - y_pred) ** 2

print("new prediction =", y_pred)
print("new loss =", loss)
```

## Training the neuron for multiple iterations

Now we repeat the forward pass, loss calculation, backpropagation,
and gradient descent update multiple times.

The goal is to see whether the loss decreases during training.

```python
x1 = 1
x2 = 2
y = 1

w1 = 0.5
w2 = -0.2
b = 0.1

learning_rate = 0.1

loss_history = []
```

```python
for epoch in range(20):

    # Forward pass
    z = w1 * x1 + w2 * x2 + b
    y_pred = relu(z)

    # Loss
    loss = 0.5 * (y - y_pred) ** 2
    loss_history.append(loss)

    # Backpropagation
    dL_dy = y_pred - y
    dy_dz = 1 if z > 0 else 0

    grad_w1 = dL_dy * dy_dz * x1
    grad_w2 = dL_dy * dy_dz * x2
    grad_b = dL_dy * dy_dz

    # Gradient descent
    w1 = w1 - learning_rate * grad_w1
    w2 = w2 - learning_rate * grad_w2
    b = b - learning_rate * grad_b

    print(
        "Epoch:", epoch + 1,
        "Prediction:", round(y_pred, 4),
        "Loss:", round(loss, 6)
    )
```

```python
import matplotlib.pyplot as plt

plt.plot(loss_history)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")

plt.show()
```

```python
print("Final w1 =", w1)
print("Final w2 =", w2)
print("Final bias =", b)
print("Final prediction =", y_pred)
print("Final loss =", loss)
```
